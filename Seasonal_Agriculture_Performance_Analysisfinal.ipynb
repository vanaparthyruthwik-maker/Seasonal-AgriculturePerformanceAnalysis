```python
from scipy import stats
```

```python
kharif_yield = df[
    df['Season'] == 'Kharif'
]['Yield_Tonnes_Ha'].dropna()

rabi_yield = df[
    df['Season'] == 'Rabi'
]['Yield_Tonnes_Ha'].dropna()

zaid_yield = df[
    df['Season'] == 'Zaid'
]['Yield_Tonnes_Ha'].dropna()

print("Kharif records:", len(kharif_yield))
print("Rabi records:", len(rabi_yield))
print("Zaid records:", len(zaid_yield))
```

```python
anova_result = stats.f_oneway(
    kharif_yield,
    rabi_yield,
    zaid_yield
)

print("ANOVA Statistic:", anova_result.statistic)
print("P-value:", anova_result.pvalue)
```

```python
alpha = 0.05

if anova_result.pvalue < alpha:
    print("Result: Statistically significant difference exists among seasonal mean yields.")
    print("We reject the null hypothesis (H0).")
else:
    print("Result: No statistically significant difference was detected among seasonal mean yields.")
    print("We fail to reject the null hypothesis (H0).")
```

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd

tukey_result = pairwise_tukeyhsd(
    endog=df['Yield_Tonnes_Ha'],
    groups=df['Season'],
    alpha=0.05
)

print(tukey_result)
```

```python
Q1 = df['Yield_Tonnes_Ha'].quantile(0.25)
Q3 = df['Yield_Tonnes_Ha'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

yield_outliers = df[
    (df['Yield_Tonnes_Ha'] < lower_limit) |
    (df['Yield_Tonnes_Ha'] > upper_limit)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Number of Yield Outliers:", len(yield_outliers))
```

```python
from scipy.stats import pearsonr

numeric_columns = df.select_dtypes(include=np.number).columns

results = []

for column in numeric_columns:
    if column != 'Yield_Tonnes_Ha':

        temp = df[[column, 'Yield_Tonnes_Ha']].dropna()

        if len(temp) >= 3:
            r, p = pearsonr(
                temp[column],
                temp['Yield_Tonnes_Ha']
            )

            results.append({
                'Variable': column,
                'Correlation': r,
                'P_Value': p
            })

correlation_significance = pd.DataFrame(results)

correlation_significance['Significant'] = (
    correlation_significance['P_Value'] < 0.05
)

correlation_significance = correlation_significance.sort_values(
    by='Correlation',
    key=abs,
    ascending=False
)

display(correlation_significance.round(4))
```

```python
significant_relationships = correlation_significance[
    correlation_significance['Significant'] == True
]

print("Statistically Significant Relationships with Yield:")
display(significant_relationships.round(4))
```

```python
season_results = []

for season in df['Season'].dropna().unique():

    season_df = df[df['Season'] == season]

    numeric_columns = season_df.select_dtypes(
        include=np.number
    ).columns

    for column in numeric_columns:

        if column != 'Yield_Tonnes_Ha':

            temp = season_df[
                [column, 'Yield_Tonnes_Ha']
            ].dropna()

            if len(temp) >= 3:

                r, p = pearsonr(
                    temp[column],
                    temp['Yield_Tonnes_Ha']
                )

                season_results.append({
                    'Season': season,
                    'Variable': column,
                    'Correlation': r,
                    'P_Value': p
                })

season_correlation_results = pd.DataFrame(season_results)

season_correlation_results['Significant'] = (
    season_correlation_results['P_Value'] < 0.05
)

display(
    season_correlation_results
    .sort_values(
        ['Season', 'P_Value']
    )
    .round(4)
)
```
